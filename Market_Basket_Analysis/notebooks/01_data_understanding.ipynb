## 4. Data Understanding — Key Metrics

Find:
- Total number of records
- Total number of columns
- Number of unique customers
- Number of unique products
- Number of unique invoices
- Date range
- Number of countries

```python
# Total number of records and columns
print(f"Total number of records : {df.shape[0]}")
print(f"Total number of columns : {df.shape[1]}")
```

```python
# Number of unique customers
print(f"Number of unique customers : {df['CustomerID'].nunique()}")
```

```python
# Number of unique products (by StockCode and Description)
print(f"Number of unique products (StockCode)   : {df['StockCode'].nunique()}")
print(f"Number of unique products (Description)  : {df['Description'].nunique()}")
```

```python
# Number of unique invoices
print(f"Number of unique invoices : {df['InvoiceNo'].nunique()}")
```

```python
# Date range
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
print(f"Date range : {df['InvoiceDate'].min()} to {df['InvoiceDate'].max()}")
```

```python
# Number of countries
print(f"Number of countries : {df['Country'].nunique()}")
```

```python
# Missing values
print("Missing values per column:")
print("-" * 35)
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Missing Count': missing, 'Percentage (%)': missing_pct})
print(missing_df)
print(f"\nTotal missing values : {df.isnull().sum().sum()}")
```

```python
# Duplicate records
dup_count = df.duplicated().sum()
print(f"Duplicate records : {dup_count}")
print(f"Percentage        : {round(dup_count / len(df) * 100, 2)}%")
```

## 6. Column Descriptions

| Column | Data Type | Description |
|---|---|---|
| **InvoiceNo** | object | Unique 6-digit invoice number; prefix 'C' indicates a cancellation |
| **StockCode** | object | Unique 5-digit product code |
| **Description** | object | Product name |
| **Quantity** | int64 | Quantity per transaction (negative = cancellation) |
| **InvoiceDate** | datetime | Date and time of the transaction |
| **UnitPrice** | float64 | Unit price in GBP (£) |
| **CustomerID** | float64 | Unique 5-digit customer identifier |
| **Country** | object | Country of the customer |

## 7. Key Observations

1. **Large dataset** — 1,041,909 transaction records across 8 columns.
2. **Missing CustomerID** — 263,308 records (~25.3%) have no CustomerID. These will need to be handled during preprocessing.
3. **Missing Description** — 2,793 records have no product description.
4. **Duplicates** — 5,268 exact duplicate rows exist in the dataset.
5. **Date Range** — Transactions span from December 2010 to December 2012 (approx. 2 years).
6. **Global Reach** — Customers from 38 countries, primarily United Kingdom.
7. **Cancelled Orders** — InvoiceNo starting with 'C' represents cancelled transactions; negative Quantity values also indicate returns/cancellations.
